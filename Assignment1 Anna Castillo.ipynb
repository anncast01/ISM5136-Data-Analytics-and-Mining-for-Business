Your task is to code a decision tree model training program in Python following the functional framework
provided below. You will receive full credits if you can implement all the functions (or similar functionality)
listed below including the main function. You will receive partial credits for each individually implemented
function. 

```python
import math
import pandas as pd

df = pd.read_csv('DT_coding_sample_training.csv')

#Function createBranches(att)
def createBranches(att, dataframe=df):
#Finds possible values of a given attribute column
    branch_list = []
    unique_values = dataframe[att].unique()
    for val in unique_values:
        branch_list.append(val)
    return branch_list

#Function: divideByLabel(path_df)
def divideByLabel(path_df):
#Finds breakdown of observations under a given path in terms of class labels
    label_counts = {}
    #Malicius is target columns
    target_series = path_df['malicious']

    for label in target_series:
        if label in label_counts:
            label_counts[label] = 1
   
    return label_counts
    
#Function: calcEntropy(aDict)
def calcEntropy(aDict):
#Calculates the entropy of a given label breakdown dictionary
    total_count = 0
    for key in aDict:
        total_count = total_count + aDict[key]

    if total_count == 0:
        return 0.0

    entropy = 0.0
    for key in aDict:
        probability = aDict[key]/total_count
        if probability > 0:
            entropy = entropy - (probability * math.log2(probability))
    
    return entropy 

#Function: findSplitAtt(current_df, attList)
def findSplitAtt(current_df, attList):
#Finds which attribute to use for split at each iteration based on highest Information Gain
    total_label_counts = divideByLabel(current_df)
    total_entropy = calcEntropy(total_label_counts)
    total_rows = len(current_df)

    best_gain = -1.0
    best_att = None

    for att in attList:
        branches = createBranches(att, current_df)
        conditional_entropy = 0.0

    for branch in branches:
        subset = current_df[current_df[att] == branch]
        subset_counts = divideByLabel(subset)
        subset_entropy = calcEntropy(subset_counts)

        weight = len(subset) / total_rows
        conditional_entropy = conditional_entropy + (weight*subset_entropy)
    
    gain = total_entropy - conditional_entropy

    if gain > best_gain:
        best_gain = gain
        best_att = att

    return best_att

#Function: checkStopping(current_df, attList)
def checkStopping(current_df, attList):
#check if a path growth needs to stop or not
    label_counts = divideByLabel(current_df)
    # 1. All records belong to the same class
    if len(label_counts) == 1:
        return "STOP: Pure node"
    
    #2. No remaining attributes left for more partitioning
    if len(attList) == 0:
        return "STOP: No attributes"
    
    #3. No records left in that partition
    if len(current_df) == 0:
        return "STOP: No records"

    return "Continue"

#Function: main(df)
def main(dataframe):
#main program
    target_column = 'malicious'
    initial_attributes = []
    for col in dataframe.columns:
        if col != target_column:
            initial_attributes.append(col)
    
    tree_paths = []

    def build_tree(current_df, attList, current_path):
        stopping_status = checkStopping(current_df, attList)
        
        if stopping_status != "Continue":
            tree_paths.append(current_path)
            return
            
        best_att = findSplitAtt(current_df, attList)
        branches = createBranches(best_att, current_df)

        remaining_atts = []
        for att in attList:
            if att != best_att:
                remaining_atts.append(att)

        for branch_val in branches:
            next_df = current_df[current_df[best_att] == branch_val]
            new_path = current_path + [(best_att, branch_val)]
            build_tree(next_df, remaining_atts, new_path)

    build_tree(dataframe, initial_attributes, [])
    return tree_paths
```

```python
tree_paths = main(df)
print("Final Decision Tree Paths:")
print(tree_paths)
index = 1
for path in tree_paths:
    print("Path", index, ":", path)
    index = index + 1
```

```python
import math

def calc_sigmoid(z):
    return 1 / (1 + math.exp(-z))

#Input Values
X1 = 0.2
X2 =0.5

#Weights
W13 = 0.05
W14 = -0.01
W15 = 0.02
W23 = 0.01
W24 = 0.03
W25 = -0.01
W36 = 0.01
W46 = 0.05
W56 = 0.015

#Biases
theta3 = -0.3
theta4 = 0.2
theta5 = 0.05
theta6 = -0.015

#Linear input and activation for internal nodes 3, 4, and 5
Z3 = (X1 * W13) + (X2 * W23) + theta3
A3 = calc_sigmoid(Z3)

Z4 = (X1 * W14) + (X2 * W24) + theta4
A4 = calc_sigmoid(Z4)

Z5 = (X1 * W15) + (X2 * W25) + theta5
A5 = calc_sigmoid(Z5)

#Linear input and activation for output node 6
Z6 = (A3 * W36) + (A4 * W46) + (A5 * W56) + theta6
A6 = calc_sigmoid(Z6)

print("Linear Input Node 3:", Z3)
print("Activation Node 3:", A3)
print("Linear Input Node 4:", Z4)
print("Activation Node 4:", A4)
print("Linear Input Node 5:", Z5)
print("Activation Node 5:", A5)
print("Linear Input Node 6:", Z6)
print("Final Activation Node 6:", A6)
```
