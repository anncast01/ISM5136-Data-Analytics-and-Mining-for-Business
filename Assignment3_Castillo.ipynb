#### 1. Function: standardizer(arr)
▪ Purpose: This function takes an array (e.g., Numpy array) of numerical observations as an input and returns the standardized (or normalized) version of the array.  

```python
import numpy as np
def standardizer(arr):
    means = np.mean(arr, axis=0)
    stds = np.std(arr, axis=0)
    stdArr = (arr - means) / stds
    return stdArr
```

#### 2. Function: calculateDistance(obj1,obj2) 
▪ Purpose: Calculates the Euclidean distance between the two objects (objects can be individual observations or centroids). You can define each object using a 1-D array (i.e., vector). 

```python
def calculateDistance(obj1, obj2):
    total = 0
    for i in range(len(obj1)):
        total = total + (obj1[i] - obj2[i]) ** 2
    distance = np.sqrt(total)
    return distance
```

#### 3. Function: findClosestCentroid(distArray) 
▪ Purpose: Given an object’s distance to all centroids, returns the index of the centroid that has the smallest distance to the object. 

```python
def findClosestCentroid(distArray):
    closestIndex = 0
    smallestDist = distArray[0]
    for i in range(len(distArray)):
        if distArray[i] < smallestDist:
            smallestDist = distArray[i]
            closestIndex = i
    return closestIndex
```

#### 4. Function: calculateCentroids(assignmentArr) 
▪ Purpose: Given a list of objects with the same cluster assignment, finds the centroid of the assignment group. Input is an n by m array, where n is the number of objects assigned to a particular cluster and m is the number of attributes.  

```python
def calculateCentroids(assignmentArr):
    centroid = np.mean(assignmentArr, axis=0)
    return centroid
```

#### 5. Function: main() 
▪ Purpose: Main program that follows the k-means pseudocode in the lecture slides. 

```python
def main():
    data = np.loadtxt("k-means_question_sample_data.csv", delimiter=",")
    n = data.shape[0]
    m = data.shape[1]
    print("Number of observations:", n)
    print("Number of attributes:", m)

    #Standardization of data 
    data = standardizer(data)

    #Step 1 - number of clusters (K)
    k = int(input("Enter the number of clusters (K): "))

    numRuns = 10
    best_SSE = -1

    for run in range(numRuns):
 
        #Step 2 - K centroids at random
        randomRows = np.random.choice(n, k, replace=False)
        centroids = data[randomRows]
     
        #No observation has a cluster yet
        assignments = np.full(n, -1)
     
        changed = True
        iteration = 0
     
        #Steps 3-5 until no cluster assignment changes
        while changed == True:
            iteration = iteration + 1
            changed = False
     
            #Steps 3 to 4 - calculate distances and assign each item to nearest centroid
            for i in range(n):
                distArray = np.zeros(k)
                for j in range(k):
                    distArray[j] = calculateDistance(data[i], centroids[j])
                closest = findClosestCentroid(distArray)
                if closest != assignments[i]:
                    assignments[i] = closest
                    changed = True
         
            #Step 5 - move centroid to mean of assigned items
            for j in range(k):
                assignment_Arr = data[assignments == j]
                if len(assignment_Arr) > 0:   # keep the old centroid if a cluster is empty
                    centroids[j] = calculateCentroids(assignment_Arr)
         
        #Sum of Squared Error (SSE)
        sse = 0
        for i in range(n):
            sse = sse + calculateDistance(data[i], centroids[assignments[i]]) ** 2
    
        print("Run", run + 1, "- initial centroids:", randomRows, "- converged after", iteration, "iterations - SSE:", round(sse, 4))
        if best_SSE == -1 or sse < best_SSE:
            best_SSE = sse
            best_centroids = centroids
            best_assignments = assignments
    
    centroids = best_centroids
    assignments = best_assignments
    sse = best_SSE


    #Final Results
    print(" ")
    print("Results (best run)")
    for j in range(k):
        print(" ")
        print("Cluster", j + 1)
        print("Number of observations:", np.sum(assignments == j))
        print("Centroid (standardized):", np.round(centroids[j], 3))
    print(" ")
    print("Cluster assignment of each observation:")
    print(assignments + 1)
    print(" ")
    print("SSE:", round(sse, 4))


main()
```
